**A. Representasi Matematis Keadaan Qubit dan Probabilitas Pengukuran**

1. Ruang Keadaan dan Basis Komputasional\
Sistem kuantum dua tingkat (qubit) direpresentasikan dalam ruang Hilbert kompleks berdimensi dua, $\mathcal{H} \cong \mathbb{C}^2$. Keadaan dasar dari sistem ini secara universal didefinisikan menggunakan basis komputasional ortonormal:$$\vert{}0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad \vert{}1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

2. Keadaan Superposisi dan Amplitudo Probabilitas\
Sebuah qubit pada kondisi awal dapat berada dalam kombinasi linear dari basis komputasional tersebut, yang diformulasikan sebagai:$$\vert{}\psi\rangle = \alpha\vert{}0\rangle + \beta\vert{}1\rangle$$di mana koefisien $\alpha, \beta \in \mathbb{C}$ bertindak sebagai amplitudo probabilitas. Selama eksistensi amplitudo $\alpha$ dan $\beta$ tidak bernilai nol, sistem komputasi tersebut terdefinisi berada di dalam keadaan superposisi.

3. Normalisasi Vektor Keadaan (L2 Norm)\
Untuk mematuhi aksioma probabilitas di mana peluang total dari seluruh keadaan harus bernilai kesatuan, vektor keadaan $\vert{}\psi\rangle$ wajib dinormalisasi. Konstanta normalisasi $\kappa$ didefinisikan menggunakan invers dari L2 Norm (norma Euclidean):$$\kappa = \frac{1}{\sqrt{\vert{}\alpha\vert{}^2 + \vert{}\beta\vert{}^2}}$$Faktor ini kemudian dikalikan pada koefisien awal untuk mendefinisikan amplitudo yang sah, yaitu $\alpha' = \kappa\alpha$ dan $\beta' = \kappa\beta$, sehingga menjamin pemenuhan kondisi $\vert{}\alpha'\vert{}^2 + \vert{}\beta'\vert{}^2 = 1$.

4. Dekohorensi dan Aturan Born (Born Rule)\
Berdasarkan Aturan Born, intervensi observasi atau pengukuran pada sistem tertutup akan memaksa fungsi gelombang mengalami dekohorensi (kolaps) ke salah satu vektor basis. Besaran probabilitas sistem untuk kolaps ke dalam keadaan $\vert{}0\rangle$ atau $\vert{}1\rangle$ diberikan oleh kuadrat dari modulus amplitudo yang ternormalisasi:$$P(0) = \vert{}\alpha'\vert{}^2, \quad P(1) = \vert{}\beta'\vert{}^2$$Dalam konteks implementasi rekayasa perangkat lunak, probabilitas absolut ini dikonversi menjadi persentase dengan faktor pengali 100, sehingga merepresentasikan peluang deterministik pada output akhir pembacaan sistem komputasi kuantum.

```python
import numpy as np

ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

alpha = 3 + 2j
beta = 1 - 2j
#   MENORMALISASIKAN AMPLITUDO GUNA MENENTUKAN PROBABILITAS
def normalize_state(alpha, beta):
    norm = np.sqrt(np.abs(alpha)**2 + np.abs(beta)**2)
    k = 1 / norm
    a_aksen = alpha * k
    b_aksen = beta * k
    return a_aksen, b_aksen

def probability(alpha, beta):
    prob_0 = np.round(np.abs(alpha)**2 * 100)
    prob_1 = np.round(np.abs(beta)**2 * 100)
    return f'State |0>: {prob_0}%, State |1>: {prob_1}%'

a_aksen, b_aksen = normalize_state(alpha, beta)
print(probability(a_aksen, b_aksen))
```

**Representasi Matematis Produk Dalam (Inner Product) dan Sifat Ortonormalitas pada Ruang Keadaan Kuantum**

1. Transformasi Konjugat Kompleks dan Notasi Bra\
Dalam mekanika kuantum, keadaan sistem direpresentasikan sebagai vektor kolom yang disebut Ket $\vert{}\psi\rangle$. Untuk melakukan evaluasi aljabar seperti produk dalam (inner product), vektor ini wajib ditransformasikan menjadi pasangan dualnya, yakni vektor baris yang disebut Bra $\langle\psi\vert{}$. Transformasi ini mengharuskan operasi Conjugate Transpose (Transpose Hermitian), di mana matriks diubah orientasinya dan setiap komponen imajiner di dalamnya dibalik tandanya ($i$ menjadi $-i$). Operasi konjugasi ini bersifat mutlak untuk memastikan bahwa produk dalam antara sebuah keadaan dengan dirinya sendiri ($\langle\psi\vert{}\psi\rangle$) selalu menghasilkan probabilitas yang berupa bilangan riil positif, mencegah terjadinya anomali nilai probabilitas negatif yang melanggar hukum fisika alam.

2. Produk Dalam (Inner Product) sebagai Amplitudo Transisi\
Operasi produk dalam yang mempertemukan Bra dan Ket, dinotasikan sebagai $\langle\phi\vert{}\psi\rangle$, bertindak sebagai operator matematis untuk menghitung derajat tumpang tindih (overlap) antara dua keadaan kuantum di dalam ruang Hilbert. Hasil skalar dari operasi ini didefinisikan sebagai amplitudo transisi. Berdasarkan Aturan Born (Born Rule), nilai absolut dari amplitudo transisi yang dikuadratkan ($\vert{}\langle\phi\vert{}\psi\rangle\vert{}^2$) merepresentasikan besaran probabilitas deterministik dari keadaan awal $\vert{}\psi\rangle$ untuk mengalami dekohorensi (kolaps) menjadi keadaan target $\vert{}\phi\rangle$ pada saat observasi dilakukan.

3. Pemenuhan Aksioma Basis Ortonormal\
Sistem koordinat atau basis komputasional $\{\vert{}0\rangle, \vert{}1\rangle\}$ yang digunakan untuk mendefinisikan ruang keadaan qubit diwajibkan memenuhi syarat Ortonormalitas, yang terdiri dari dua kriteria absolut:\
**Kondisi Normal (Panjang Kesatuan):** Evaluasi basis terhadap dirinya sendiri harus merepresentasikan peluang utuh (100%), yang dibuktikan secara aljabar melalui $\langle0\vert{}0\rangle = 1$ dan $\langle1\vert{}1\rangle = 1$.\
**Kondisi Ortogonal (Saling Lepas):** Evaluasi basis terhadap basis independen lainnya harus menunjukkan ketiadaan irisan probabilitas sama sekali, dibuktikan melalui $\langle0\vert{}1\rangle = 0$ dan $\langle1\vert{}0\rangle = 0$.

4. Mekanisme Proyeksi dan Ekstraksi Amplitudo\
Terdapat segregasi fundamental antara vektor basis dengan amplitudo probabilitas. Vektor basis ($\vert{}0\rangle$ dan $\vert{}1\rangle$) merupakan matriks konstanta absolut yang bertindak sebagai kerangka ruang koordinat stasioner. Sebaliknya, koefisien amplitudo probabilitas ($\alpha$ dan $\beta$) merupakan nilai skalar kompleks yang mendefinisikan keadaan spesifik partikel di dalam ruang tersebut. Operasi produk dalam terhadap salah satu basis, seperti $\langle0\vert{}\psi\rangle$, secara mekanis berfungsi sebagai operator proyeksi aljabar (fungsi ekstraksi) untuk mengisolasi nilai amplitudo.Hal ini dibuktikan melalui penjabaran distribusi:$$\langle 0 \vert{} \psi \rangle = \langle 0 \vert{} (\alpha\vert{}0\rangle + \beta\vert{}1\rangle)$$$$\langle 0 \vert{} \psi \rangle = \alpha\langle 0 \vert{} 0 \rangle + \beta\langle 0 \vert{} 1 \rangle$$$$\langle 0 \vert{} \psi \rangle = \alpha(1) + \beta(0)$$$$\langle 0 \vert{} \psi \rangle = \alpha$$Substitusi sifat ortonormal pada persamaan di atas menjamin bahwa nilai amplitudo $\alpha$ pada keadaan superposisi dapat diekstrak secara presisi tanpa kontaminasi komputasional dari koefisien amplitudo basis lainnya.

5. Dualitas Konsep Normalisasi: Kondisi Aksiomatik dan Mekanisme Transformasi\
Dalam tinjauan aljabar linear untuk sistem kuantum, terminologi "normalisasi" merepresentasikan dua entitas matematis yang berbeda namun terikat secara kausalitas. Evaluasi $\langle\psi\vert{}\psi\rangle = 1$ berkedudukan sebagai syarat mutlak (aksioma validasi) yang mengevaluasi apakah probabilitas total dari sebuah keadaan kuantum valid secara fisika. Sebaliknya, perkalian skalar menggunakan konstanta normalisasi $\kappa$ berkedudukan sebagai mekanisme transformasi linier untuk memperbaiki keadaan mentah yang belum ternormalisasi ($\vert{}\psi_{\text{raw}}\rangle$). Distribusi skalar $\kappa$ terhadap vektor tersebut ($\kappa\alpha$ dan $\kappa\beta$) secara matematis menjamin bahwa vektor hasil akhir ($\vert{}\psi'\rangle$) akan lolos dari aksioma validasi. Hal ini dibuktikan melalui substitusi matriks:$$\langle\psi'\vert{}\psi'\rangle = (\kappa\langle\psi_{\text{raw}}\vert{})(\kappa\vert{}\psi_{\text{raw}}\rangle)$$$$\langle\psi'\vert{}\psi'\rangle = \kappa^2 \langle\psi_{\text{raw}}\vert{}\psi_{\text{raw}}\rangle$$Dengan mendefinisikan $\kappa = \frac{1}{\sqrt{\langle\psi_{\text{raw}}\vert{}\psi_{\text{raw}}\rangle}}$, persamaan tersebut tereduksi menjadi:$$\langle\psi'\vert{}\psi'\rangle = \left( \frac{1}{\sqrt{\langle\psi_{\text{raw}}\vert{}\psi_{\text{raw}}\rangle}} \right)^2 \langle\psi_{\text{raw}}\vert{}\psi_{\text{raw}}\rangle$$$$\langle\psi'\vert{}\psi'\rangle = \frac{\langle\psi_{\text{raw}}\vert{}\psi_{\text{raw}}\rangle}{\langle\psi_{\text{raw}}\vert{}\psi_{\text{raw}}\rangle} = 1$$Pembuktian ini memverifikasi bahwa normalisasi amplitudo melalui skalar konstan secara ekuivalen adalah metode mutlak untuk memenuhi hukum kekekalan probabilitas pada ruang Hilbert.

**Mekanisme Pengukuran Kuantum: Transformasi Amplitudo Kompleks Menjadi Output Biner Klasik Melalui Pemodelan Stokastik**

1. **Pemetaan Ruang Kompleks ke Ruang Probabilitas (Aplikasi Aturan Born)**\
Dalam mekanika kuantum, keadaan suatu sistem sebelum pengukuran dipostulatkan berada dalam kondisi superposisi koheren, direpresentasikan oleh vektor keadaan $\vert{}\psi\rangle$ yang berdimensi $N$. Vektor ini memuat koefisien amplitudo probabilitas $c_k$ yang berwujud bilangan kompleks ($\mathbb{C}$).Proses observasi memicu kolapsnya keadaan tersebut (dekohorensi). Fungsi simulasi pengukuran pertama-tama bertindak sebagai operator pemetaan matematis yang mengekstrak magnitudo dari setiap amplitudo berdasarkan Aturan Born (Born Rule). Ekstraksi ini mentransformasikan vektor kompleks menjadi distribusi probabilitas riil ($\mathbb{R}$):$$P(k) = \vert{}c_k\vert{}^2$$Di mana aksioma total probabilitas dipastikan ekuivalen dengan $1$ ($\sum \vert{}c_k\vert{}^2 = 1$), yang membuktikan bahwa sistem berada pada keadaan yang telah ternormalisasi sempurna.

2. **Pemodelan Pengukuran via Variabel Acak Diskrit (Discrete Random Variable)**\
Karena komputasi kuantum tunduk pada prinsip probabilistik alam, hasil akhir observasi tidak dapat diprediksi secara deterministik tunggal. Secara statistika, instrumen pengukuran direpresentasikan sebagai Variabel Acak Diskrit $X$.\
Ruang Sampel ($\Omega$): Didefinisikan oleh seluruh indeks keadaan basis yang mungkin terjadi, yakni $\Omega = \{0, 1, \dots, N-1\}$.\
Fungsi Massa Probabilitas (PMF): Probabilitas sistem untuk kolaps pada suatu keadaan klasik tertentu $k$ didefinisikan secara presisi oleh bobot hasil turunan Aturan Born, yakni $P(X = k) = \vert{}c_k\vert{}^2$.Dalam implementasi komputasional, model ini dieksekusi menggunakan algoritma weighted random sampling (pengambilan sampel acak berbobot) yang mendistribusikan kemungkinan hasil tarikan secara proporsional terhadap luasan amplitudo kuadratnya.

3. **Hukum Bilangan Besar dan Konvergensi Statistik (Shots/Measurements)**\
Satu kali pengukuran pada partikel fisik akan secara permanen merusak vektor keadaan $\vert{}\psi\rangle$, menyisakan hanya satu buah informasi skalar (misal: 0 atau 1). Oleh karena itu, probabilitas analitis awal tidak dapat diobservasi secara langsung.Untuk merekonstruksi distribusi probabilitas secara empiris, algoritma mensimulasikan pengukuran independen secara berulang (dikenal sebagai shots atau $M$). Berdasarkan Hukum Bilangan Besar (Law of Large Numbers), seiring dengan bertambahnya jumlah observasi independen ($M \to \infty$), frekuensi relatif kemunculan state $k$ pada himpunan sampel akan berkonvergensi mendekati probabilitas teoritis mutlaknya:$$\lim_{M \to \infty} \frac{n_k}{M} = \vert{}c_k\vert{}^2$$Simulasi eksekusi berulang ini mutlak diperlukan untuk menguji ketahanan (robustness) sebuah algoritma kuantum terhadap varians statistik yang muncul akibat keterbatasan sampel empiris di dunia nyata.

4. **Demarkasi Realitas Fisik dan Destruksi Informasi Koheren**\
Fungsi simulasi pengukuran (measurement function) pada dasarnya berperan sebagai "Filter Realita" yang membatasi akses pengamat (CPU Klasik) terhadap black-box ruang kuantum (QPU/Perangkat Keras Kuantum).Meskipun simulator perangkat lunak secara memori memiliki akses penuh terhadap vektor amplitudo mentah, fungsi pengukuran dirancang secara ketat untuk mendestruksi informasi koheren tersebut pada tahap akhir komputasi. Keluaran (API Response) dari fungsi ini diwajibkan hanya meretur deretan data skalar berwujud array biner (kumpulan bilangan bulat klasik). Pembatasan terdesain ini meniru perilaku fisik Quantum Processing Unit (QPU) sungguhan, sehingga memaksa arsitektur post-processing program hulu (seperti backend penyortir atau frontend analitik) untuk beradaptasi murni dari manipulasi data empiris biner, bukan dari bocoran analitik matriks ket itu sendiri.
